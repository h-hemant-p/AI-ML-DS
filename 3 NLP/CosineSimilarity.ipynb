# Cosine Similarity

Cosine similarity measures **how similar two vectors are based on their direction (angle), not magnitude**.

* 👉 Think of vectors like arrows:
   - Same direction → highly similar
   - 90° → unrelated
   - Opposite → completely different

* **Mathematical Formula**

$\cos(\theta) = \frac{A \cdot B}{|A| |B|}$

* Where:
   - **A · B** → dot product
   - **||A||, ||B||** → magnitudes (lengths)
   - **θ (theta)** → angle between vectors

* **👉 Output range**
   - **+1 → exactly same**
   - **0 → no similarity**
   - **-1 → opposite**



* **Cases**

| Angle (θ) | Cosine Value | Meaning   |
| --------- | ------------ | --------- |
| 0°        | 1            | Identical |
| 90°       | 0            | Unrelated |
| 180°      | -1           | Opposite  |

👉 Smaller angle = higher similarity

---

* **Intuition (VERY IMPORTANT)**

👉 Cosine similarity ignores magnitude:

Example:
* A = [100, 200]
* B = [1, 2]

👉 Direction same → similarity = **1**

✔️ This is why it's perfect for:
* Text comparison
* NLP
* High-dimensional data

---

**Step-by-Step Calculation**

Example:

A = [3, 2, 0, 5]
B = [1, 0, 0, 0]

1. Step 1: Dot product

```
(3×1 + 2×0 + 0×0 + 5×0) = 3
```

2. Step 2: Magnitudes

```
||A|| = √(3² + 2² + 0² + 5²) = 6.16
||B|| = √(1²) = 1
```

3. Step 3: Final

```
Cosine = 3 / (6.16 × 1) = 0.49
```

👉 Moderate similarity
