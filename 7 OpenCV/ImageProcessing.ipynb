# Canny Edge Detection

* Use of Canny Edge detection 
    * to Detect Borders (Face borders etc.)
    * to Saperate Objects
    * for Feature Extraction
    * Sketch effect dalne ke liye kisi bhi image me

```python
'''
edges = cv2.Canny(image, threshold1=100, threshold2=200)
thresold1 = lower boundry to detect weak edges
thresold2 = upper boundry to detect strong edges
every pixel have 2 options either it is edge(complete black) or not edge(complete white)
if pixel value is greater than thresold2 then it is strong edge
if pixel value is less than thresold1 then it is not edge
if pixel value is between thresold1 and thresold2 then it is weak edge
if weak edge is connected to strong edge then it is considered as edge otherwise not edge
'''

img = cv2.imread('image.png', cv2.IMREAD_GRAYSCALE) # image must be in grayscale
edges = cv2.Canny(img, 50, 200)

cv2.imshow('Image', img)
cv2.imshow('Edges', edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Bitwise Operations

* Use of Bitwise Opertions
    * combine 2 images
    * 1 Images me se kuchh part cut out karna hai using another image wo bhi kar sakte hain
    * flip/remove specific areas of image

* Must be
    * Image1 and Image2 height, width same
    * Use only black and white image
    * 

```python
img1 = np.zeros((300, 300), dtype="uint8")
img2 = np.zeros((300, 300), dtype="uint8")

cv2.rectangle(img1, (100, 100), (250, 250), 255, -1)
cv2.circle(img2, (150, 150), 100, 255, -1)

# cv2.imshow('Rectangle', img1)
# cv2.imshow('Circle', img2)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

bitwise_and = cv2.bitwise_and(img1, img2)
bitwise_or = cv2.bitwise_or(img1, img2)
# bitwise_xor = cv2.bitwise_xor(img1, img2)
bitwise_not = cv2.bitwise_not(img1)

cv2.imshow('AND', bitwise_and)
cv2.imshow('OR', bitwise_or)
# cv2.imshow('XOR', bitwise_xor)
cv2.imshow('NOT', bitwise_not)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
