```python
import cv2 

image = cv2.imread('image.png')
```

##### **Gaussian Bluring**
Isme kernel size ke according matrix bante hain on top of image and matrix ke sare elements ka avg nikal ke usko matrix ke center wale element se replace kar dete hain. isaliye karnel size jitna bada hoga image utni hi blur hogi. And kernel size hamesa odd hi honi chahiye.

```python
# (kernel_size_x, kernel_size_y) = must be odd numbers

image_kernel_size = (15, 15)
standard_deviation_in_x_direction = 2
blurred_image = cv2.GaussianBlur(image, image_kernel_size,standard_deviation_in_x_direction)
cv2.imwrite("blurred_image.png", blurred_image)
cv2.imshow("Blurred Image", blurred_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

##### Median Blur
Isame ham avg value na nikal ke middle value nikalte hain of neigours(matrix sare elements ka middle and usko replace kar denge matrix ke center element se)

```python
blur_image = cv2.imread('blurred_image.png')

if blur_image is None:
    print("Error: Could not load image.")
    exit()

cleaned_image = cv2.medianBlur(blur_image, 5)
cv2.imshow("Cleaned Image", cleaned_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### Sharpning

```python
import numpy as np
image = cv2.imread('image.png')
if image is None:
    print("Error: Could not load image.")
    exit()

sharpening_kernel = np.array([
    [0, -1, 0],
    [-1, 5,-1],
    [0, -1, 0]
    ])

depth = -1
sharpened_image = cv2.filter2D(image, depth, sharpening_kernel) # -1 means the output image has the same depth as the source image

cv2.imwrite("sharpened_image.png", sharpened_image)
cv2.imshow("Sharpened Image", sharpened_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
