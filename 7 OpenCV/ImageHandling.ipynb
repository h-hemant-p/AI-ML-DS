```python
import cv2
```

```python
"""flag is optional 
if flag = 1, it reads the image in color mode
if flag = 0, it reads the image in grayscale(black and white) mode
if flag = -1, it reads the image including the alpha(unchanged) channel
"""
flag = 1
image = cv2.imread("image.png", flag)
if image is not None:
    print("Image loaded successfully.")
    cv2.imshow("Image", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

    # Save the image
    cv2.imwrite("image_copy.png", image)
    print("Image saved successfully.")
```

##### Image Shape

```python
if image is not None:
    h, w, channels = image.shape
    print(f"Image Height: {h}, Width: {w}, Channels: {channels}")
```

##### Colored to Gray Image

```python
if image is not None:
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    cv2.imshow("Grayscale Image", gray_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

##### Image Resizing

```python
if image is not None:
    resized_image = cv2.resize(image, (300, 300))
    cv2.imshow("Resized Image", resized_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    # cv2.imwrite("resized_image.png", resized_image)
```

##### Image Cropping

```python
if image is not None:
    cropped_image = image[150:900, 200:900] # start_y:end_y, start_x:end_x
    cv2.imshow("Cropped Image", cropped_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

##### Image Rotation 

```python
if image is not None:
    (h,w) = image.shape[:2] # using slice to get height and width value only
    center = (w // 2, h // 2) 
    M = cv2.getRotationMatrix2D(center, 45, 1.0) # center, angle, scale
    rotated_image = cv2.warpAffine(image, M, (w, h))
    cv2.imshow("Rotated Image by 45 degrees", rotated_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

##### Image Flipping

```python
if image is not None:
    flipped_horizontal = cv2.flip(image, 1) # 1 for horizontal flip
    flipped_vertical = cv2.flip(image, 0)   # 0 for vertical flip
    flipped_both = cv2.flip(image, -1)      # -1 for both horizontal and vertical flip
    cv2.imshow("Original Image", image)
    cv2.imshow("Flipped Horizontally", flipped_horizontal)
    cv2.imshow("Flipped Vertically", flipped_vertical)
    cv2.imshow("Flipped Both", flipped_both)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

##### Draw a Line

```python
if image is not None:
    lined_image  = cv2.line(image.copy(), (0,0), (250,250), (255,0,0), 5) # image, start_point, end_point, color(BGR), thickness
    cv2.imshow("Lined Image", lined_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

##### Draw a Ractangle

```python
if image is not None:
    recangled_image = cv2.rectangle(image.copy(),(0,0), (250,250), (255,0,0), 3) # image, start_point, end_point, color(BGR), thickness
    cv2.imshow("Rectangled Image", recangled_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

##### Draw a Circle

```python
if image is not None:
    circled_image = cv2.circle(image.copy(), (250,250), 100, (0,255,0), 3) # image, center_coordinates, radius, color(BGR), thickness
    cv2.imshow("Circled Image", circled_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

##### Adding Text

```python
if image is not None:
    texted_image = cv2.putText(image.copy(), "OpenCV Demo", (50,50), cv2.FONT_HERSHEY_SIMPLEX, 1, (0,0,255), 2) # image, text, orgin(coordinates), font, fontScale, color(BGR), thickness
    cv2.imshow("Texted Image", texted_image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
