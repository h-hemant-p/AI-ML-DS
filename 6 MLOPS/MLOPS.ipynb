# What is MLOPS?

MLOPS or Machine learning Operations, is essentially devops for machine learning, Just like devops manages end to end life cycle of machine learning models.

This includes everything from:
*   Data collection and preprocessing
*   Model training
*   Testing and validation
*   Deployment to production
*   Continous monittoring and improvement

-> The goal to make the model lifecycle auditable, reproducible and automated.

-> Think of a typical ML project: a data scientist trains a model on their laptop or in a jupyter notebook.
It works well there, but deploying it to a live application- serving predictions to thousands of users - required version control scalability,testing and monitoring.

MLOPs provides the practices, tools and processes to bridge this gap.


# Why do we need MLOPS?

MLOPS helps to:

Deploy,monitor,update and manage ML models smoothly just like devops does for software.

But the journey from a trained model to production ready model is full of challenges:

1. Deployment issues: Moving a model from development to production can fail due to dependencies, incompatible environments or lack of testing.

-> Data mismatch: Training data is not equal production data, model accuracy drops because the data changes.

2. Monitor issues: Once deployed, models can degrade over time as data patterns change(called data drift)

-> Data drift: Data in prduction chnages slowly, but alert triggered-> model becomes inaccurate.

3. Lifecycle management: It means that managing the entire journey of a machine learning model from the moment you build it -> deploy it -> monitor it-> update it -> retire it.


# Process of MLOPS

The MLOPS processes is typically divided into three majot phases: Design, Model development, Operations.

1. Design phase: This design phase is all about understanding the business problem and planning its solution.

Key activities includes:

*   Gathering requirements from stakeholders.
*   Defining ML use cases and prioritizing them.
*   Assessing the data availability and quality.
*   Specifying the fucntional and non fucntional requirements.

2. Model development:

*  Perform data engineering(cleaning, feature extraction)
*  Experiment with different algorithms
*  Train and validate models
*  Test model performance

3. Operations phase: It is the phase where the model is running in production and you continously monitor, maintain, update it.

What happens in the operations phase?

-> Monitoring the model: It means that continuously checking how the model is performing after the deployment.

*   Track accuracy
*   Check data drift
*   Check the latency
*   Check prediction errors
*   Detect concept drift

All three phases are interconnected.

Design decisions impact experimentation and experimentation result guide deplyment starategies.


# Understanding ML Pipelines

To make this process repeatable and scalable, we use ML pipelines

The ML pipeline is a sequence of automated steps- data preparation, model training, evaluation and deployment.

Pipeline components are like bulding blocks:
*   Each components performs a single task(eg: data preprocessing, model training).
*   Components are portable and reusable.

# Hyperparameter Tuning:

*  Before deployment we need to optimize the models perfromance.
*  This involves hyperparameters whihc are setting that govern the training process- like learning rate, batch size or number of decision tree.

Hyperparameter tuning involves:
*  Running multiple training jobs with hyperparameter combination.
*  Evaluating results to find the best configuration.
