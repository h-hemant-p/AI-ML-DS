Step 1: Create a resource group

*   Resource group is a container in cloud platform that holds and organizes your resoureces.

Eg: It is like a folder where you keep all related cloud resouces together.

- If you create a web application, you may put all these inside one resource group:

*   Virtual machine
*   Database
*   Storage
*   Network

Step 2: You need to search for azure machine learning to create a the workspace.

An Azure machine learning workspace is like the main control room for all our ML work.


Step 3: Go to resource group in that there is an option of azure machine learning studio, click on that and then we will be re-directing to some new page

Step 4: Select Compute Instance

COmpute refers to the virtual machine or servers that do the actual work like:

1. Training machine learning models
2. Running the python scripts
3. Deploying web apps
4. Processing data

Step 5: After creating the compute go to assets in that select data, give any name to it and make sure that you are keeping the type as tabular.


Step 6: After uploading the data go to automated ML.
