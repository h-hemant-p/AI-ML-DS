
### **Correlation**

* Correlation measures the **linear relationship** between two numerical variables.
* It works **only on numerical data** — never on strings or categorical data.
* The correlation value ranges from **–1 to +1**:

  * **+1** → perfect positive linear relationship
  * **–1** → perfect negative linear relationship
  * **0** → no linear relationship
* Correlation shows **relationship**, not **causation** — so it has limited use in decision-making.
* Correlation is **sensitive to outliers**; one extreme value can change the result.
* It cannot detect **non-linear relationships**, so it is not very useful for complex datasets.
* In ML, correlation is mainly used for:

  * **Feature selection** (removing weakly correlated features)
  * **Checking multicollinearity** (identifying highly correlated features)

* **Using the Formula (Pearson Correlation)**
  * The most common way is **Pearson correlation**:
  * $r = \frac{\text{Cov}(X, Y)}{\sigma_X \cdot \sigma_Y}$
    * Where:
      * Cov(X, Y) = covariance of X and Y
      * σX = standard deviation of X
      * σY = standard deviation of Y
  * But in real life you rarely calculate this manually.


* Heatmaps quickly show strong/weak correlations.


```python
import pandas as pd

df = pd.DataFrame({
    "x": [1, 2, 3, 4],
    "y": [2, 4, 5, 8]
})

df.corr()   # gives correlation matrix

# or directly
# df['x'].corr(df['y'])
```
