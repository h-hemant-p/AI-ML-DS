## Activation Function

* An **activation function** is a mathematical function used in **neural networks** to introduce **non-linearity** into the model.
* **Activation Function = Adds Non-Linearity + Controls Output Range + Enables Learning of Complex Patterns**
* Without activation functions, a neural network would behave like a  linear regression model , no matter how many layers it has.

* Activation function Introduces non-linearity into the model that allows the model to learn complex relationships in data.
* Controls output range – e.g., 0–1 or -1–1.
* Enables backpropagation – makes gradient-based learning possible.
* Adds flexibility – each layer can learn different representations

* **Activations Functions**

    | Type                                | Formula                                                 | Output Range  | Notes                                                                           |
    | ----------------------------------- | ------------------------------------------------------- | ------------- | ------------------------------------------------------------------------------- |
    | **1. Sigmoid**                      | $ f(x) = \frac{1}{1 + e^{-x}} $                         | (0, 1)        | Smooth curve; used in binary classification; can cause **vanishing gradients**. |
    | **2. Tanh (Hyperbolic Tangent)**    | $ f(x) = \tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}} $ | (-1, 1)       | Better than sigmoid; still can suffer from vanishing gradient.                  |
    | **3. ReLU (Rectified Linear Unit)** | $ f(x) = \max(0, x) $                                   | [0, ∞)        | Most used; fast and simple; can cause **“dead neurons”** if output stays 0.     |
    | **4. Leaky ReLU**                   | $ f(x) = x \text{ if } x > 0, \text{ else } 0.01x $     | (-∞, ∞)       | Fixes ReLU’s dead neuron issue.                                                 |
    | **5. Softmax**                      | $ f(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}} $             | (0, 1), sum=1 | Used in **output layer** for **multi-class classification**.                    |
    | **6. Swish (by Google)**            | $ f(x) = x \cdot \sigma(x) $                            | (-∞, ∞)       | Smooth and often outperforms ReLU in deep models.                               |


* **Where to Use**

    | Layer                               | Common Activation       |
    | ----------------------------------- | ----------------------- |
    | Hidden Layers                       | ReLU, Leaky ReLU, Swish |
    | Output (Binary Classification)      | Sigmoid                 |
    | Output (Multi-Class Classification) | Softmax                 |
    | Output (Regression)                 | None / Linear           |
