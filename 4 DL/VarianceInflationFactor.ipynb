# **VIF (Variance Inflation Factor)**


* VIF is a measure used to detect **multicollinearity** in regression models.

* **Multicollinearity = when two or more independent variables are highly correlated with each other.**

* This causes:
    * Unstable coefficients
    * High variance
    * Wrong predictions

* So VIF tells you **how much the variance of a coefficient is inflated** due to multicollinearity.



* **Formula**
    * $\text{VIF} = \frac{1}{1 - R^2}$
    * Where:
        * **R²** = how much one variable can be predicted using other independent variables.


* **Interpretation of VIF**

    | VIF Value | Meaning                                             |
    | --------- | --------------------------------------------------- |
    | **1**     | No multicollinearity                                |
    | **1 – 5** | Moderate correlation, OK to use                     |
    | **> 5**   | High multicollinearity, try to fix                  |
    | **> 10**  | Very high multicollinearity → remove/merge variable |


* **Why is VIF Important?**
    * Because high VIF means:
        * Model becomes unstable
        * Coefficients become unreliable
        * Predictions can change drastically with small changes in data
    * So VIF helps you clean your feature set.

* **How to Fix High VIF?**
    * Remove one of the correlated features
    * Combine them (PCA, Feature Engineering)
    * Keep one and drop the others
    * Use regularization (Ridge, Lasso)

* **VIF(Variance Inflation Factor)**
    * **VIF measures how much multicollinearity is present in a regression model.**
    * It shows **how much the variance of a coefficient is inflated** because of correlation with other independent variables.

    * **Formula**
    * $VIF = \frac{1}{1 - R_j^2}$
    * Where:
        * $( R_j^2 )$ = R-squared from a regression of one independent variable **against all the other independent variables**.

* **How to interpret VIF**
    * **VIF = 1** → No multicollinearity (perfect).
    * **VIF < 5** → Acceptable.
    * **VIF > 5** → Moderate multicollinearity (warning).
    * **VIF > 10** → High multicollinearity (serious problem).

* **Why is VIF used?**
    * To check if any feature is highly correlated with others, which can:
        * make model coefficients unstable,
        * increase variance,
        * reduce model reliability.

* **How to Reduce High VIF**
    1. **Remove highly correlated features**
        * If two features give the same information, remove one.
        * Example:
            * *House size (sq ft)*
            * *Number of rooms*
                * Choose only one → VIF drops.

    2. **Combine correlated features**
        * Use **feature engineering**:
        * Make a **ratio**, **average**, or **index** to combine them.
        * This reduces redundancy.

    3. **Use Dimensionality Reduction (PCA)**
        * Principal Component Analysis removes correlation by creating new uncorrelated components.

    4. **Standardize or Normalize features**
        * This sometimes reduces VIF in models sensitive to scale (like ridge regression).

    5. **Use L1 or L2 Regularization**
        * **L1 (Lasso)** → removes irrelevant features
        * **L2 (Ridge)** → reduces coefficient variance
            * This helps stabilize multicollinearity.

    5. **Add more data**
        * More data → more stable coefficients → lower VIF.

    7. **Drop categorical variables with too many levels**
        * High-cardinality categories can also cause high VIF.
        * Reduce or group them.
