# Nural Networks

* A Neural Network (NN) is a model inspired by how the human brain works.
* It is made up of layers of neurons (nodes) that process data and learn patterns from it.


---


* **Epoch** : 
    - an epoch is one complete pass of the entire training dataset through the model.
    - Imagine you have 1,000 training samples
    - You feed all of them once into the model → that’s 1 epoch
    - Example
        - Dataset size = 1,000 samples
        - Batch size = 100
        - Then:
            - 1 epoch = 10 iterations
            - If you train for 5 epochs → model sees the data 5 times
