

## **File Formats to Save Machine Learning Models**

* **1. `.pkl` (Pickle File)**
    * Used for saving:
        * Trained ML models (Sklearn)
        * **Tokenized values**
        * **Standardized values**
        * **Normalized values**
        * Any **preprocessing objects** (encoders, scalers, vectorizers)
    * Best for **traditional ML models** and preprocessing pipelines.
    * Preferred when you want to save **all small operations** together.

* **2. `.h5` (HDF5 — Hierarchical Data Format)**
    * Commonly used for **Deep Learning models**, especially **Keras / TensorFlow**.
    * Stores:
        * Model architecture
        * Weights
        * Optimizer state
        * Supports large and complex models with multiple layers.
        * Good for **cross-platform** and **huge model** storage.

* **3. `.keras`**
    * Newer format introduced by Keras.
    * Saves:
        * Architecture
        * Weights
        * Training configuration
        * More stable and preferred in **latest TensorFlow/Keras versions**.
        * Recommended replacement for `.h5` for future compatibility.


* **Summary**

| Format     | Best For                          | Notes                                |
| ---------- | --------------------------------- | ------------------------------------ |
| **.pkl**   | ML models + preprocessing objects | Most preferred for sklearn pipelines |
| **.h5**    | Deep learning (Keras/TensorFlow)  | Stores entire model + weights        |
| **.keras** | Latest Keras models               | Recommended modern format            |
