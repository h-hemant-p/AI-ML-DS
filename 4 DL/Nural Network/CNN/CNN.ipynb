# Convulational Nural Network

* Limitations of ANN
    1. ANNs cannot handle images well
        * ANN treats every pixel as an independent feature.
        * Images have thousands of pixels → input layer becomes huge → model becomes inefficient.
        * For a 224×224×3 (244px, 3rgb) image, ANN needs 150,528 weights in just the first layer.
