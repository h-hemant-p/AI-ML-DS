## **Deep Learning**

Imagine you have a robot teacher. You give it your math homework:

* At first, it guesses randomly (lots of mistakes)
* Every time it's wrong, you say "No, that's wrong the answer should be 10, not 8",
* The robot teacher fixes its "thinking switches" a little.
* After many tries, it leaarns to get mmost answers correct.

Thats Deep Learning -> Learning from mistakes until it gets realy good.

* **Machine Learning vs Deep Learning**
    1. ML
        * Machine Learning is like teaching a kid with **rules and hints**.
        * Example : You want a kid to recognize fruits.
        * You Say : "Apples are usually red, round, and sweet. Bananas are yellow and long."
        * The kid uses those **given rules** to guess.
        * So in ML, we give the computer **data + features (hints),** it learns patterns.
        * In short : **ML = Learn with Guidence(Features are hand made by us)**
    2. DL
        * Deep Learning is like giving the kid a lot of fruits **without telling the rules**.
        * You just say: "Here's 1,000 apples and 1,000 bananas figure it out yourself."
        * The kid's brain notices:
            1. First layer → "Oh, red and round often go together."
            2. Next layer → "Oh, yellow and long often go together."
        * Eventually, the kid can say “That's an apple!" without you ever telling the rules.
        * In short: DL = Learning the rules by itself (no hand-made feature

* **Why Deep Learning is More Popular Now?**
    1. *Data Explosion*
        * In the past, we didn't have enough data for deep networks to be useful.
        * Now we have massive amounts of data: images (Instagram, YouTube), text (social media, books, articles), and speech (calls, podcasts).
        * Deep Learning thrives on big data → the more it gets, the smarter it becomes.
    2. *Hardware Power*
        * In the past, we didn't have enough computing power to train deep networks.
        * Now we have massive amounts of computing power: GPUs (Graphics Processing Units), TPUs (Tensor Processing Units), and ASICs (Application-Specific Integrated Circuits).
        * Deep Learning thrives on big computing power → the more it gets, the smarter it becomes.
    3. *Automatic Feature Learning*
        * In Machine Learning, we had to manually choose features (like "length of email" to detect spam).
        * Deep Learning learns features automatically from raw pixels, raw text, sound.
        * This makes it powerful for unstructured data.
    4. *Better Performance*
        * Deep Learning models consistently beat traditional Machine Learning models in accuracy for tasks like:
            * Image Recognition (Photos, Selfies, Videos)
            * Speech Recognition (Calls, Podcasts)
            * Natural Language Processing (Chatbots, Translation)
        * Example: Google translate beacome more better after switching from ML to Deep Learning.
    5. *Versatility Across Domains*
        * Work on around every type of data: text, image, audio, video, time series etc.
        * Same "Deep Networkd Idea" powers
            * Self-driving cars
            * Chatgpt 
            * Face recognition
            * Netflix recommendations
            * Medical image analysis
    6. *Reaserch Breakthroughs*
        * Innovations like CNNs, RNNs, LSTMs, Transformers, GANs, etc. have made deep learning super flexible and powerful.

* **Perceptron**
    * The Perceptron is the simplest type of artificial neuron, introduced by Frank Rosenblatt in 1958. It is the foundation of neural network and Deep learning.
    * It mimics how a biological neuron works taking inputs, processing them, and giving an output.
    * When we are using one Output and one Input layer then its a perceptron but in between if you use any hidden layer then it becomes ANN.
