# K-Means 

K-Means is an unsupervised machine learning algorithm used to group similar data points into K clusters.

* "K" = number of clusters (user-defined)
* K is an Hyperparameter
* Goal = minimize distance within clusters, maximize distance between clusters

* How to Find the Optimal Value of K in K-Means
    * The goal is to choose a value of K (number of clusters) such that adding more clusters does not significantly improve the model.
    * We will find the best value of K using Elbow method.
---

1. Step 1 : Choose K
    * Decide number of clusters (K)
2. Step 2 : Initialize Centroids
    * Randomly pick K points as initial cluster centers
3. Step 3 : Assign Points to Clusters 
    * For each(ALL) data point:
        - Calculate distance to all centroids
        - Assign to nearest centroid
4. Step 4: Update Centroids
    * Recalculate centroid = mean of all points in that cluster
5. Step 5: Repeat
    * Repeat Step 3 & 4 until:
        - Centroids don’t change OR
        - Max iterations reached
6. Step 6: Final Clusters
    * You get:
        - K clusters
        - Each with its centroid

---
* **Total Within-Cluster Variation(TWCV)**
    - Measures how compact your clusters are (how close data points are to their cluster center)
    - If points are very close to centroid → variation is low ✅ (good clustering)
    - If points are far from centroid → variation is high ❌ (bad clustering)
---
* n_init = (Hyperparameter)number of times the K-means algorithm is run with different initial centroids
* Why do we need it?
    - K-means has a problem:
    - It depends on random initialization of centroids
    - Bad initialization → bad clusters ❌
    - Good initialization → good clusters ✅

* What n_init does
    - Instead of running K-means once:
    - It runs K-means multiple times (n_init times)
    - Each time with different random centroids
    - Then: It selects the result with the lowest within-cluster variation (WCSS / inertia)

* Note : K will remain the same for every run

---

* **Step-by-Step Process (Elbow Method)**
    1. Start with K = 1
        - Run K-Means clustering.
        - Compute Within-Cluster Variation (WCV).
    2. Increase K gradually (K = 2, 3, 4, …)
        - For each value of K:
            - Apply K-Means.
            - Calculate Within-Cluster Variation (WCV).
    3. Plot the graph
        - X-axis → Number of clusters (K)
        - Y-axis → Within-Cluster Variation (WCV)
    4. Analyze the graph
        - WCV decreases as K increases.
        - Initially, the decrease is sharp.
        - After a certain point, the decrease becomes slow.
    5. Choose optimal K
        - Select the value of K at the “elbow point”
        - This is where the rate of decrease sharply changes (bend in the graph).

**WCSS(Withing Cluster Sum of Square)**

* WCSS calculates the sum of the squared [Euclidean distances](https://opendatascience.com/unsupervised-learning-evaluating-clusters/) between each data point in a cluster and that cluster's centroid (the mean center point).

* Low WCSS: Indicates that the data points are very close to their centroids, resulting in tight, well-defined clusters.
* High WCSS: Indicates that data points are spread far from their centroids, suggesting the clusters are loose or poorly defined.

* For $k$ clusters, WCSS is expressed as:
    * $WCSS = \sum_{i=1}^{k} \sum_{j=1}^{n_i} \text{dist}(x_j^{(i)}, c_i)^2$
    * Where:
        * $k$: Total number of clusters.
        * $x_j^{(i)}$: The $j^{th}$ data point in cluster $i$.
        * $c_i$: The centroid of cluster $i$. [8, 9] 

* Monotonic Decrease: As you increase the number of clusters ($k$), WCSS will always decrease. This is because more clusters allow centroids to be closer to individual data points. If $k$ equals the total number of data points, WCSS becomes zero.
* Inertia: In popular libraries like [Scikit-learn](https://jtemporal.com/kmeans-and-elbow-method/), WCSS is referred to as Inertia.
---
* For a cluster $S_i$ containing $n$ data points, the centroid $c_i$ is calculated as:
    * $c_i = \frac{1}{n} \sum_{j=1}^{n} x_j$ 
    * Where:
        * $c_i$: The centroid of cluster $i$.
        * $n$: The total number of data points in that cluster.
        * $x_j$: Each individual data point (vector) within the cluster.
