* Types of Clustering
    1. Execlusive Clustering: Each data point can only be a member of one of the k clusters that you are farming.
    2. Overlapping Clustering: Each data point should be the momber of every cluster that you are farming
        - Data point holdes the some degree of membership in each cluster. and sum of all degrees will be 1.
        - The algorithm that performs overlapping clustering will compute the degrees of membership.
        - For example **Fuzzy or C-Means Clustering**
    3. Hierarchical Clustering
        1. Step 1: 
            - Based in Euclidian distance we identify the 2 most similar data points and merge them in a single cluster, 
            - if 2 or more distances are exactly same and also lowest then also merge them and create new clusters
            - also now we can give the name to the cluster and the data points that are alone yet they also will be considered as cluster but with a single member
        2. Step 2:
            - Now Calculate the centroid the of all clusters
            - Then we calculate the same distance between all clusters centroids
            - and this distance between centroids is also know as inter-cluster distance
            - Now we will merge the clusters based on lowest distance
        3. Step 3:
            - Reapeat the step 2 jab tak sare points khatm nhi ho jate
            - At the end sare cluster 1 single cluster ke andar aa jayenge

# K-Means 

K-Means is an unsupervised machine learning algorithm used to group similar data points into K clusters.

* "K" = number of clusters (user-defined)
* K is an Hyperparameter
* Goal = minimize distance within clusters, maximize distance between clusters
