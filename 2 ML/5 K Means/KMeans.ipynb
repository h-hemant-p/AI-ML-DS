# K-Means 

K-Means is an unsupervised machine learning algorithm used to group similar data points into K clusters.

* "K" = number of clusters (user-defined)
* K is an Hyperparameter
* Goal = minimize distance within clusters, maximize distance between clusters


1. Step 1 : Choose K
    * Decide number of clusters (K)
2. Step 2 : Initialize Centroids
    * Randomly pick K points as initial cluster centers
3. Step 3 : Assign Points to Clusters
    * For each(ALL) data point:
        - Calculate distance to all centroids
        - Assign to nearest centroid
4. Step 4: Update Centroids
    * Recalculate centroid = mean of all points in that cluster
5. Step 5: Repeat
    * Repeat Step 3 & 4 until:
        - Centroids don’t change OR
        - Max iterations reached
6. Step 6: Final Clusters
    * You get:
        - K clusters
        - Each with its centroid

---
* **Total Within-Cluster Variation(TWCV)**
    - Measures how compact your clusters are (how close data points are to their cluster center)
    - If points are very close to centroid → variation is low ✅ (good clustering)
    - If points are far from centroid → variation is high ❌ (bad clustering)
---
* n_init = (Hyperparameter)number of times the K-means algorithm is run with different initial centroids
* Why do we need it?
    - K-means has a problem:
    - It depends on random initialization of centroids
    - Bad initialization → bad clusters ❌
    - Good initialization → good clusters ✅

* What n_init does
    - Instead of running K-means once:
    - It runs K-means multiple times (n_init times)
    - Each time with different random centroids
    - Then: It selects the result with the lowest within-cluster variation (WCSS / inertia)

* Note : K will remain the same for every run
