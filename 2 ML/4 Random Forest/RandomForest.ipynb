# Random Forest

* Random forest is based on decision tree.
* Means It can be used in creating both classification model and regration model just like decision tree.
* Random Forest is an ensemble machine learning algorithm.
* It combines multiple Decision Trees to improve performance.
* Works for:
    - Classification
    - Regression
* Idea: Instead of relying on one tree → use many trees → take combined decision.
* The number of Base models in a Random Forest is a Hyper Parameter.
* Base learner, Base model, base Estimator - referes to the individual model in ensemble algorithm
* Consolidating balearner prediction
    - Majority Voting (If problem is an classification problem)
    - Averaging (If problem is an regration problem)
* Every input should be feeding to all all base models

* Random Forest uses bagging to construct ensembles of randomized decision trees
* **Bagging** :  Bootstrap Sampling and aggregation (together called Bagging).
* Uses Bootstrap Sampling to derive multiple new datasets from one initial training dataset to train multiple base models.
* Because if we will trail all base models with same dataset then all models performance will be same that we dont want.

* **Bootstrap Sampling**
    * Bootstrap Sampling is a method where we:
        - Randomly select data points from a dataset
        - With replacement
        - 👉 With replacement means:
            - The same data point can be selected multiple times
            - Some data points may not be selected at all
    * Size of sample = size of original dataset
    * Around ~63% unique data is used per sample and Rest are duplicates
    * Some points repeated, Some points left out
    * 👉 The left-out data is called:🔸 Out-of-Bag (OOB) Data (Used for model validation)
