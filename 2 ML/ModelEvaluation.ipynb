## **Confusion Matrix**

The confusion matrix is a matrix used to determine the performance of the classification models for a given set of test data.
A confusion matrix is a tabular representation used to show the actual vs predicted classifications of a model.



| Actual \ Predicted | Positive | Negative |
| ------------------ | -------- | -------- |
| Positive           | TP       | FN       |
| Negative           | FP       | TN       |


* **True Negative(TN):** Model has given prediction No, and the real or actual value was also No.
* **True Positive(TP):** The model has predicted yes, and the actual value was also true.
* **False Negative(FN):** The model has predicted no, but the actual value was Yes, it is also called as  **Type-II error** .
* **False Positive(FP):** The model has predicted Yes, but the actual value was No. It is also called a **Type-I error.**

* **Purpose**
  1. Shows how many predictions are correct or incorrect
  2. Helps identify Type-I and Type-II errors
  3. Acts as the base for calculating performance metrics

* Output Type
  1. Raw counts / numbers
  2. No score or percentage

* **Example Use**
  1. Understanding where the model is making mistakes
  2. Medical diagnosis error analysis

# Performance Matrices

* In Machine Learning, performance metrics are quantitative measures used to evaluate how well a model is performing on a given task. They tell you how accurate, reliable, or useful your model is — not just whether it works.
* Performance metrics are numerical measures that evaluate how well a model performs, calculated using the confusion matrix.
* Performance metrics are derived from the confusion matrix.

* Why Performance Metrics Are Important
    1. Compare different ML models
    2. Detect overfitting or underfitting
    3. Choose the best model for production
    4. Measure real-world impact (especially with imbalanced data)

* **Purpose**
    1. Quantifies model performance
    2. Allows comparison between models
    3. Helps decide model suitability for production

* **Output Type**
    * Percentages / ratios / scores

* **Example Use**
    * Model selection
    * Performance reporting
    * Handling imbalanced datasets

---
* **Classification Accuracy:** It is one of the important parameters to determine the accuracy of the classification problems.
    * $\text{Accuracy} = \frac{TP + TN}{TP + TN + FP + FN}$

---
* **Misclassification rate:** the error rate is a metric that quantifies the proportion of incorrect predictions made by a classification model.
    * $\text{Error Rate} = \frac{FP + FN}{TP + TN + FP + FN}$

---
* **Precision:** It can be defined as the number of correct outputs provided by the model or out of all positive classes that have predicted correctly by the model, how many of them were actually true.
    * $\text{Precision} = \frac{TP}{TP + FP}$

---
* **Recall:** It is defined as the out of total positive classes, how our model predicted correctly. The recall must be as high as possible.
    * $\text{Recall} = \frac{TP}{TP + FN}$

---
* **F1-Score:** If two models have low precision and high recall or vice versa, it is difficult to compare these models. So, for this purpose, we can use F1-score. This score helps us to evaluate the recall and precision at the same time. The F1-score is maximum if the recall is equal to the precision.
    * $\text{F1-Score} = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}$

---
* **R2 Score :**
    * The R² score is a performance metric used in regression problems
    * R² tells us how much of the output variation is explained by the model.

> Fromula :  $R^2 = 1 - \frac{SS_{res}}{SS_{tot}}$

* Where:
    * $(SS_{res} = \sum (y - \hat{y})^2)$ → Residual sum of squares
    * $(SS_{tot} = \sum (y - \bar{y})^2)$ → Total sum of squares
    * $(y)$ = actual value
    * $(\hat{y})$ = predicted value
    * $(\bar{y})$ = mean of actual values

* **📊 Range of R² Score**

    | R² Value | Meaning                    |
    | -------- | -------------------------- |
    | **1.0**  | Perfect prediction         |
    | **0.8**  | 80% variance explained     |
    | **0.0**  | Model performs like mean   |
    | **< 0**  | Worse than mean prediction |


* **Example**
    * If:
        * Actual house prices vary a lot
        * Your model explains **85%** of that variation
    * Then:
        * > **R² = 0.85**

* **❌ Limitations of R²**
    * Does **not** show prediction error magnitude
    * Can be misleading with **overfitting**
    * Always increases when more features are added
> Use **Adjusted R²** to fix this issue.


* **Adjusted R² (Better Version)**

$ Adjusted\ R^2 = 1 - \left(\frac{1 - R^2}{n - p - 1}\right)(n - 1) $

* Where:
    * (n) = number of observations
    * (p) = number of features


> **R² score measures how much variance in the dependent variable is explained by the regression model.**

# Overfitting and Underfitting

* **Overfitting**
    * In training, the model performs very well (low bias).
    * In testing, the model performs poorly due to high variance.
    * ✔ Model learns training data too well (including noise)
* **Underfitting**  
    * In training, the model performs poorly due to high bias.
    * In testing, the model also performs poorly.
    * ✔ Model is too simple to learn patterns

| Case         | Training Performance | Testing Performance | Bias | Variance |
| ------------ | -------------------- | ------------------- | ---- | -------- |
| Underfitting | Poor                 | Poor                | High | Low      |
| Overfitting  | Very Good            | Poor                | Low  | High     |
| Good Fit     | Good                 | Good                | Low  | Low      |

# **Cost Function/Loss Function**

* A cost function (also called loss function) measures how well or poorly your model is performing.
* Cost ffunction is the average of all *Residual errors*
* **all Residual error** : the difference between an *observed value* and a *predicted value*
    * $y_i - (a_1x_i + a_0)$
* Sare errors ko sum karke total number of outputs se divide karte hain
* **MSE :** 
    * It calculates the **average of squared differences** between the **actual values** and the **predicted values** — that’s why it’s called **Mean Squared Error (MSE)**.
    * $MSE = \frac{1}{n} \sum_{i=1}^{n} \left(y_i - (a_1x_i + a_0)\right)^2$
        * $ n $: number of data points
        * $ y_i $: actual (true) value
        * $ x_i $: input feature value
        * $ a_1 $: slope (weight or coefficient)
        * $ a_0 $: intercept (bias)
        * $ (a_1x_i + a_0) $: predicted value
        * $ y_i - (a_1x_i + a_0) $: prediction error
        * The **square** ensures all errors are positive and penalizes large ones.

* Sometimes, it’s also written as a **cost function** with a factor of ½ for easier differentiation:
    * $J(a_0, a_1) = \frac{1}{2n} \sum_{i=1}^{n} (y_i - (a_1x_i + a_0))^2$
    * The ½ in the cost function is added purely for mathematical convenience — it simplifies gradient descent calculations but does not change the behavior or results of the model.
* Both forms are correct —
    * MSE → for performance measure
    * J → for optimization in training

* We have to minimize the value of cost function by changeing the value of $a_0$ and  $a_1$

# **Gradient descent**
* Gradient Descent is an optimization algorithm used to minimize the cost function (or loss function)
* Working of Gradient Descent
    * Initialize random values for slope and intercept
    * Evaluate the error between the actual output and the predicted output
    * Identify how much each parameter contributes to the error.
    * Optimize the parameters in the direction that minimizes the error
    * Iterate till the error minimizes as much as possible.

* **Best Notes :** [https://www.tpointtech.com/linear-regression-in-machine-learning](https://www.tpointtech.com/linear-regression-in-machine-learning)

* **Global minima :** global minima wo point hota hai jaha loss function ki value sabse kam ho jati hai matlab model sabse achha perform karta hai aur predection me sabse kam galtiyan hoti hain.(Lowest value in gradient descent curve)
* **Local Minima :** minima point of a pertucular region.
---

* **Convergence theorem (Repeat convergence theorem):** This theorem states that when we use an optimization algorithm like gradient descent, the model’s weights gradually move towards a stable value(Global Minima), minimizing the loss function. Essentially, it ensures that the model reaches a stable solution where further updates don’t significantly change the outcome.
---


* **Learning Rate :** The learning rate is a parameter that controls how big the update step should be when adjusting the model’s parameters during gradient descent.
    * If it’s too **high**, the model may skip the optimal point.
    * If it’s too **low**, training becomes very slow.
    * Learning rate high rahega to loss function ki value kafi jyada different hongi aur global minima find karne me problem aayegi.
---

* **Hypothesis Testing :** Jab ham (80% data training ke liye and 20% data testing ke liye) train test split karte hain aur training ke bad test data se model testing karte usi ko hypothesis testing bolte hain.

* **Multicolonierity** : Multicollinearity happens when two or more independent variables in a regression model are highly correlated with each other.
---


* **1. Linear Regression Hypothesis**
    * $h_\theta(x) = \theta_0 + \theta_1 x$

* **2. Loss / Cost Function (MSE)**
    * $J(\theta) = \frac{1}{2m} \sum_{i=1}^{m} (h_\theta(x^{(i)}) - y^{(i)})^2$

* **3. Gradient Descent Update Rule**
    * This is the key formula related to **convergence**:
    * $\theta_j := \theta_j - \alpha \frac{\partial J(\theta)}{\partial \theta_j}$

* **4. Convergence Condition**
    * Gradient descent **converges** when:
    * $J(\theta^{(t+1)}) \approx J(\theta^{(t)})$
    * or simply:
    * $|\theta^{(t+1)} - \theta^{(t)}| < \epsilon$
    * (where **ε** is a very small number)
    * Meaning: **Updates become very small → model has reached a stable point (minimum).**
---


* **Homoscedascity**(Error Must be in range) : Homoscedasticity means that the variance of the errors (residuals) remains constant across all values of the independent variable(s).

    * In simple words:
        * 👉 The spread of errors is the same everywhere.
        * 👉 No “fan shape”, no increasing or decreasing spread.
