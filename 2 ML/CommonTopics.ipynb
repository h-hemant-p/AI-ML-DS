# Bias and Variance

* **Bias :** 
    * Bias = error due to over-simplified model
    * Happens when model makes strong assumptions about data to simplify the training
    * High bias → Model is too simple
    * Fails to capture patterns
    * Leads to Underfitting
    * Example : Using linear model for complex nonlinear data

* **Variance :**
    * Variance = error due to model being too sensitive to training data
    * Model learns noise instead of actual pattern
    * High variance → Model is too complex
    * Performs well on training but poorly on test data
    * Leads to Overfitting
    * Example:Deep decision tree with very high depth

* **Bias-Variance Tradeoff :**
    * Goal: Balance both
    * If bias ↓ → variance ↑
    * If variance ↓ → bias ↑
    * Best model = Optimal balance

* **Visual Understanding**
    * High Bias → Straight line (misses pattern)
    * High Variance → Wiggly curve (overfits noise)
    * Ideal → Smooth curve (captures true pattern)

* **Bias vs Variance Difference**
    | Feature        | Bias                           | Variance                 |
    | -------------- | ------------------------------ | ------------------------ |
    | Definition     | Error due to wrong assumptions | Error due to sensitivity |
    | Model Type     | Simple model                   | Complex model            |
    | Problem        | Underfitting                   | Overfitting              |
    | Training Error | High                           | Low                      |
    | Test Error     | High                           | High                     |
