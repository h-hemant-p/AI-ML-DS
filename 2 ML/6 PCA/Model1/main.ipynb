# Standardization

If we apply standardization in our data, mean will become 0 and standard deviation will become 1.

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
pd.DataFrame(X_scaled)
```

```python
X_scaled.std()
```

```python
X_scaled.mean()
```

We need to find the relation between features (independent columns) and gain important information. To do that we will now find the covariance matrix.

Covariance matrix help PCA understand how the data features are related, so it can reduce the dimensions while keeping the most important information.

```python
X_T = X_scaled.T
pd.DataFrame(X_T)
```

```python
cov_mat = np.cov(X_T)
pd.DataFrame(cov_mat)
```

With the help of covariance matrix now we can easily find out the eighen values and eighen vectors.

Eighen vectors tells us the best angle so that we capture the maximum information from it.

Eigen values tells us how much value we will get from the direction of eigen vectors.


```python
eig_values, eig_vectors = np.linalg.eig(cov_mat)
```

```python
pd.DataFrame(eig_values)
```

```python
pd.DataFrame(eig_vectors)
```

# Explained Variance
Explained variance tells us how much information(or variance) each principal component captures.

More variance means it is explaining principal component, more important it is.


# How to calclulate?

To find out total information: we divide the eigen values by the sum of all eigen values.


```python
pd.DataFrame(eig_values)
```

```python
total_sum=sum(eig_values)
```

```python
total_sum
```

```python
sorted(eig_values,reverse=True)
```

```python
var_exp = []

for x in sorted(eig_values,reverse=True):
  exp=(x/total_sum)*100
  var_exp.append(exp)
var_exp
```

# Cumulative variance

Inorder to find the proper principal component and explained variance we calculate cumulative variance.

It helps us to decide how many components to keep.  If we get 95% variance with first few components, you can ignore the rest while still having most of the important information.


```python
cumulative_var_exp = np.cumsum(var_exp)
cumulative_var_exp
```

```python
cumulative_df = pd.DataFrame(cumulative_var_exp)
cumulative_df.head(30)
```

```python
plt.figure(figsize=(10,5))
plt.bar(range(len(var_exp)), var_exp,label='Individual_explained_variance', color='g')
plt.step(range(len(cumulative_var_exp)), cumulative_var_exp,label='Cumulative variance explained')
plt.xlabel('Explained Variance Ratio')
plt.ylabel('Explained variance Ratio')
plt.legend()
plt.show()
```
