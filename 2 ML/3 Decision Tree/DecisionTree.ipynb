# Decision Tree

* Decision tree was originally developed for classification(Decision Tree based Classifier).
* but Decision tree concept with some modification for Regresion as well.
* means with the help of decision tree we can create both classifier and regression model.

> Graphical representation of all the possible solutions to a tree.

* A decision tree is a supervised machine learning algorithm that models decisions using a tree-like hierarchical structure of questions and outcomes. It is widely used for both classification and regression tasks because it produces interpretable models that mimic human decision-making and require minimal data preprocessing.
* 

* **Learning type:** Supervised (classification and regression)
* **Structure:** Root node, internal decision nodes, branches, and leaf nodes
* **Splitting criteria:** Information Gain, Gini Impurity, or Chi-Square
* **Notable algorithms:** ID3, C4.5, CART, CHAID
* **Applications:** credit scoring, medical diagnosis, customer segmentation, and fault detection.
* Each Leaf Node always should have a label assigned to it.
* **Advantages:**
    * Highly interpretable and handle both numerical and categorical data without scaling.
    * Performs feature selection intrinsically and are computationally efficient at prediction time.
* **Limitations:** Prone to overfitting and sensitivity to data variation

* **Entropy** : Measure of uncertinity or randomness
    - Value of Entropy can be between 0 to 1.
    - 1 Means maximum uncertinity and 0 means minimum uncertinity.
    - Entropy Formula
        - The entropy of a dataset $S$  is calculated as:
            - $H(S) = - \sum_{i=1}^{n} p_i \log_2(p_i)$
        - Where
            - **H(S)** = Entropy of dataset $S$
            - **pᵢ** = Probability of class $i$
            - **n** = Number of classes
            - **log₂** = Log base 2

* **Information Gain:** Entropy of examples(data points) before testing(splliting) the features minus the weighted average of emtropy of examples(data points) after testing(splliting) the features.
    - $InfoGain = H_{before} - H_{after}$
    - Highest InfoGain ∝ High Entropy Reduction
