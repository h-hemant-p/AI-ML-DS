## Supervised Learning

* When we have labeled data then we use supervise learning.

* **Types of Supervise Learning**
    * **Regression :**
        * Regression is a machine learning task where the goal is to predict a continuous numerical value based on input features. The output is a real number (Contineous values).
        * Real number output : Infinite (uncountable).
        * **Examples** :
            * Predicting house prices based on size, location, and number of bedrooms.
            * Forecasting sales revenue for a company.
            * Estimating a person’s age from biometric data.
        * Algorithms:
            1. Linear Regration
    * **Classification** :
        * Classification is a machine learning task where the goal is to predict the category or class label of a data point based on its features. The output is discrete (categorical).
        * Descrete output : countable values.
        * **Examples** :
            * Predicting whether an email is spam or not spam (binary classification).
            * Classifying an image as a dog, cat, or bird (multi-class classification).
            * Determining if a patient has a disease (e.g., positive or negative).
        * Algorithms:
            1. Logistic Regration
    * **Hybrid Supervised Learning Algorithms -**
        * Decision Tree
        * Random Forest
        * K Nearest Neighbors(KNN)
        * Naive Bayes
        * Support Vector Machine(SVM)


## Unsupervised Learning

* Unsupervised Learning is a type of Machine Learning where the model is trained on  **unlabeled data** .

* No predefined output labels.
* The system tries to **find hidden patterns, structures, or relationships** within the data.
* **Goal** : To organize data into groups or reduce data complexity without prior knowledge of the outcomes.

* **Types of UnSupervised Learning:**
    1. **Clustering**
        * Grouping similar data points together.
        * Examples: Customer segmentation, document classification.
        * **Algorithms** :
            * K-Means Clustering
            * Hierarchical Clusterin
            * DBSCAN (Density-Based Spatial Clustering)
    2. **Dimensionality Reduction**
        * Reducing the number of input variables/features while preserving essential information.
        * Helps in visualization and reducing computation.
        * **Techniques** :
            * PCA (Principal Component Analysis)
            * t-SNE (t-Distributed Stochastic Neighbor Embedding)
            * LDA (Linear Discriminant Analysis – semi-supervised)
            * Autoencoders (Neural Networks)
