#### Linear Regration


* Linear Regression is a supervised machine learning algorithm used to predict a continuous numerical value by fitting a straight line between input (X) and output(Y).
* Linear Regration is an Regression Model.
* Linear Regression me input features continuous aur discrete dono ho sakte hain, lekin output (target) usually continuous hota hai.

---
* **Types**
    1. **Simple Linear Regration**
        * a single independent variable is used to predict the value of a numerical dependent variable
        * **FORMULA:**  Y=MX+C
            * Y Predicted output
            * X = Input variable (feature)
            * m = Slope (kitni steep line hai)
            * c = Intercept (line Y-axis ko kaha cut karti hai)
    2. Multiple Linear Regratio
        * more than one independent variable is used to predict the value of a numerical dependent variable
        * **FORMULA:**  $ŷ= α_0+ α_1 x_1+ α_2 x_2+ ...+ α_n x_n$
            * ŷ = Predicted output (dependent variable)
            * $x_1,x_2,...,x_n$ = Input variables (independent features)
            * $α_0$ = Shows intercept(bias term)
            * $α_1, α_2,...,α_n$ = Coefficients (how much each feature affects the output)



---
* **Types of Linear Regression Line**
    1. Positive Linear Relation
    2. Negative Linear Relation
