# Data Science Notes: Data Types & Tokens (From a Data Science Perspective)

## Topic 1: Understanding Data in Data Science

In data science, data is the core element used for analysis, prediction, and model training. Data is commonly understood in two major ways:

1. By structure/dimension (Scalar, Vector, Matrix, Tensor)
2. By nature/type (Numerical, Categorical, etc.)

---

## Topic 2: Data Types Based on Dimensions (Mathematical View)

This classification is very important in machine learning, deep learning, and linear algebra.

### 1. Scalar (0D)

A scalar is a single numeric value.

**Examples in Data Science:**

* Temperature = 32
* Salary = 50000
* Loss value of a model
* Accuracy score

**Usage:**

* Model evaluation metrics
* Individual feature values

---

### 2. Vector (1D)

A vector is a list of numbers representing multiple features of one data point.

**Examples:**

* One student’s data:
  [Age, Marks, Attendance] → [20, 85, 90]

**Usage in Data Science:**

* Feature representation of a single sample
* Input to ML models

Each row in a dataset is typically treated as a vector.

---

### 3. Matrix (2D)

A matrix is a table of numbers arranged in rows and columns.

**Examples:**

* Dataset table:
  Rows → Samples
  Columns → Features

Example:

| Age | Salary | Experience |
| --- | ------ | ---------- |
| 25  | 30000  | 2          |
| 30  | 50000  | 5          |

**Usage:**

* Entire dataset representation
* Training data for ML models
* Linear algebra operations

---

### 4. Tensor (3D and Higher)

A tensor is a multi-dimensional extension of matrices.

**Examples in Data Science:**

* Color image → 3D tensor (Height × Width × Channels)
* Video → 4D tensor (Frames × Height × Width × Channels)
* Deep learning batch data → 3D/4D tensors

**Usage:**

* Deep learning (CNN, RNN, Transformers)
* Image processing
* NLP embeddings

---

## Topic 3: Dataset Structure in Data Science

In practical ML projects:

* One value → Scalar
* One record (row) → Vector
* Entire dataset → Matrix
* Deep learning input → Tensor

---

## Topic 4: Data Types Based on Nature (Statistical View)

This classification is heavily used in data analysis and feature engineering.

### 1. Numerical Data

Numbers that represent measurable quantities.

**Types:**

* Continuous → Height, Weight, Temperature
* Discrete → Number of students, Number of orders

---

### 2. Categorical Data

Labels or names.

**Examples:**

* City names
* Gender
* Product category

---

### 3. Ordinal Data

Categorical data with order.

**Examples:**

* Rating: 1, 2, 3, 4, 5
* Rank: 1st, 2nd, 3rd

---

### 4. Binary Data

Only two possible values.

**Examples:**

* Yes/No
* 0/1
* True/False

---

## Topic 5: What is a Token in Data Science (NLP Perspective)

In Natural Language Processing (NLP), text data cannot be used directly by ML models. It must be converted into smaller units called tokens.

### Definition

A token is the smallest unit of text used for processing in NLP models.

A token can be:

* A word
* Part of a word
* A character
* A punctuation mark

---

### Example

Sentence:
"I love machine learning"

Possible tokens:

* "I"
* "love"
* "machine"
* "learning"

Total = 4 tokens

But sometimes:
"learning" → "learn" + "ing"

So token count may vary.

---

## Topic 6: Why Tokens Matter in Data Science

Tokens are important in NLP because models cannot understand raw text. They understand numbers.

Process:

Text → Tokens → Numbers → Model Training

This is called:

* Tokenization
* Encoding
* Vectorization

---

## Topic 7: Tokens and Feature Engineering in NLP

In NLP projects:

* Each token becomes a feature.
* Tokens are converted into vectors using:

  * Bag of Words
  * TF-IDF
  * Word Embeddings
  * Transformer embeddings

---

## Topic 8: Tokens in LLM-Based Data Science Work

In modern AI systems:

* Input text is broken into tokens
* Tokens are converted into embeddings (vectors)
* Models process these vectors

Example:
Chatbots, RAG systems, document search, summarization.

---

## Topic 9: Quick Concept Map

Data Science Pipeline:

Raw Data → Clean Data → Features → Model

For text data:

Text → Tokens → Vectors → Model → Prediction

---

## Final Key Insight

From a data science perspective:

* Scalar = Single value
* Vector = One data sample
* Matrix = Entire dataset
* Tensor = Deep learning input
* Tokens = Units of text used in NLP models

In NLP and AI:
Text is first converted into tokens, then into vectors, and finally into tensors for model training.
