```python
import cv2

cap = cv2.VideoCapture(0) # 0 for default camera, 1 for external camera, 2 for another external camera

while True:
    ret, frame = cap.read() # ret is a boolean indicating if the frame was read correctly, frame is the image array
    if not ret:
        break

    cv2.imshow('Video Feed', frame)
    # gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    # cv2.imshow('Gray Video', gray)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

##### Video Recording

```python
camera = cv2.VideoCapture(0)

frame_width = int(camera.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(camera.get(cv2.CAP_PROP_FRAME_HEIGHT))
print(f'Frame Width: {frame_width}, Frame Height: {frame_height}')

# compression method
codec = cv2.VideoWriter_fourcc(*'XVID') # 'XVID' is a popular codec for AVI files
fps = 20 # frames per second

recorder= cv2.VideoWriter('recorded_video.avi', codec, 20.0, (frame_width, frame_height))

while True:
    success, frame = camera.read()
    if not success:
        break

    recorder.write(frame) # write the frame to the video file

    cv2.imshow('Recording Video', frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

camera.release()
recorder.release()
cv2.destroyAllWindows()
```
